```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load CSV
df = pd.read_csv(r"C:\Users\ZAK-TECH\Desktop\KAIM week 2 new\solar-challenge-week2\Scraper\notebooks\task2_with_sentiment.csv")

# Convert dates
df['review_date'] = pd.to_datetime(df['review_date'])

# For trend analysis
df['year_month'] = df['review_date'].dt.to_period('M')

# Set seaborn style
sns.set_style("whitegrid")


# -----------------------------
# 2. Average Sentiment per Bank
# -----------------------------
plt.figure(figsize=(10,5))
sns.barplot(data=df, x='bank_name', y='sentiment_score', estimator='mean', ci=None)
plt.xticks(rotation=45)
plt.title("Average Sentiment Score per Bank")
plt.ylabel("Average Sentiment Score")
plt.xlabel("Bank Name")
plt.show()

# -----------------------------
# 3. Rating Distribution
# -----------------------------
plt.figure(figsize=(6,4))
sns.histplot(df['rating'], bins=5, kde=True)
plt.title("Rating Distribution of Reviews")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.show()

# -----------------------------
# 4. Sentiment Trend Over Time
# -----------------------------
trend = df.groupby('year_month')['sentiment_score'].mean()

plt.figure(figsize=(12,5))
plt.plot(trend.index.astype(str), trend.values, marker='o')
plt.title("Monthly Average Sentiment Trend")
plt.xticks(rotation=45)
plt.ylabel("Average Sentiment Score")
plt.xlabel("Year-Month")
plt.show()

# -----------------------------
# 5. Rating vs Sentiment (Scatter)
# -----------------------------
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='rating', y='sentiment_score', hue='bank_name')
plt.title("Rating vs Sentiment Score per Bank")
plt.xlabel("Rating")
plt.ylabel("Sentiment Score")
plt.legend(title='Bank', bbox_to_anchor=(1,1))
plt.show()
```
