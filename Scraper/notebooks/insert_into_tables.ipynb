```python
#connect to the database
conn = psycopg2.connect(
    host="localhost",
    database="bank_reviews",
    user="postgres",
    password="Talieynaso1996&4"
)

cur = conn.cursor()
```

```python
conn.rollback()
```

```python

# Loop through each row in df_banks
for _, row in banks_df.iterrows():

    # Execute SQL INSERT for each row
    # ON CONFLICT DO NOTHING avoids errors if the primary key already exists
    cur.execute(
        """
        INSERT INTO banks (app_id, title, score, ratings, reviews, installs, bank_code, bank_name)
        VALUES (%s, %s, %s, %s, %s, %s, %s, %s)
        ON CONFLICT DO NOTHING;
        """,
        (
            row['app_id'],
            row['title'],
            row['score'],
            row['ratings'],
            row['reviews'],
            row['installs'],
            row['bank_code'],
            row['bank_name']
                  
        )
    )
```

```python
for _, row in reviews_df.iterrows():

    # Insert each review into PostgreSQL
    cur.execute(
        """
        INSERT INTO reviews 
            (review_id,review_text,rating,bank_code,bank_name,source)
        VALUES 
            (%s, %s, %s, %s, %s, %s)
        ON CONFLICT DO NOTHING;
        """,
        (
            row['review_id'],
            row['review_text'],
            row['rating'],
            row['bank_code'],
            row['bank_name'],
            row['source']
        )
    )

# -------------------------------------------------------
# Save changes and close connection
# -------------------------------------------------------

conn.commit()   # Saves all INSERT operations permanently
cur.close()     # Close the cursor
conn.close()    # Close the database connection

print("Data inserted successfully!")
```
